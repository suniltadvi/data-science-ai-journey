```python
# =========================================================
# PROJECT 1: STOCK PRICE ANALYZER
# You are a data scientist at a Finance company
# You have daily stock prices of a company for 10 days
# Your job: Analyse the prices using a FOR loop
# =========================================================

stock_prices = [1450, 1523, 1498, 1601, 1587,
                1623, 1549, 1678, 1701, 1689]

company_name = "RELIANCE"

print(f"Stock analysis report for {company_name}")
print("=" * 45)

# Loop through each price
for i in range(len(stock_prices)):
    price = stock_prices[i]
    day = i+1
    print(f"Day {day:2} | Price: Rs{price}")

print("=" * 45)
print(f"Total days analysed: {len(stock_prices)}")
```

```python
# =====================================================
# Finding Highest, Lowest using FOR loop
# =====================================================

# stocks purchased at various prices
stock_prices = [1450, 1523, 1498, 1601, 1587,
                1623, 1549, 1678, 1701, 1689]

# start with first price as reference
highest_price = stock_prices[0]
lowest_price = stock_prices[0]
total = 0

# loop through all prices
for price in stock_prices:
    total = total + price   #total value of stocks

    if price > highest_price:
        highest_price = price

    if price < lowest_price:
        lowest_price = price

#calculate average
average_price = total / len(stock_prices)

print("STOCK ANALYSIS SUMMARY")
print("=" * 40)
print(f"Highest Price : Rs{highest_price}")
print(f"Lowest Price: Rs{lowest_price}")
print(f"Average Price: Rs{average_price}")
print(f"Total value: Rs{total}")
print("=" * 40)

# Bonus: Flag days where price was above average
print("\n Days where price was ABOVE average:")
for i in range(len(stock_prices)):
    if (stock_prices[i]) > average_price:
        print(f"   Day {i+1}: Rs{stock_prices[i]} <--ABOVE AVERAGE")
        
```

```python
#=========================================================
# Example -2: IP Address Threat Scanner
#=========================================================
# Identify suspicious IPs from your network logs with IP addresses

# List of IPs
ip_addresses = ["192.168.1.1", "10.0.0.5", "185.234.219.80",
                "192.168.1.105", "45.33.32.156", "10.0.0.12", 
                "192.168.1.200", "103.216.221.15", "172.16.0.5",
                "10.0.0.8"]
# Known supsicious IP range (starts with these)
suspicious_ranges = ["185.", "45.", "103."]

safe_ips = []
threat_ips = []

print("CYBERSECURITY THREAT SCANNER")
print("=" * 40)
print("Scanning all IP addresses......\n")

# Loop through every IP address
for ip in ip_addresses:
    is_threat = False

    # Check if IP starts with any suspicious range
    for suspicious in suspicious_ranges:
        if ip.startswith(suspicious):
            is_threat = True

    if is_threat:
        threat_ips.append(ip)
        print(f" THREAT DETECTED : {ip}")
    else:
        safe_ips.append(ip)
        print(f" SAFE            : {ip}")

# Final Report
print("\n" + "=" * 40)
print(" THREAT SCAN REPORT")
print("=" * 40)
print(f"Total IPs Scanned : {len(ip_addresses)}")
print(f"Safe IPs        : {len(safe_ips)}")
print(f"Threats Found   : {len(threat_ips)}")
print(f"Threat Level    : ", end="")
if len(threat_ips) == 0:
    print("LOW — All Clear!")
elif len(threat_ips) <= 2:
    print("MEDIUM — Monitor closely!")
else:
    print("HIGH — Take immediate action!")

print("\n Blocked IPs:")
for threat in threat_ips:
    print(f"{threat}")
```
